```python
import numpy as np


np.random.seed(15)

matriz_latencia = np.random.uniform(
    5,
    80,
    (6, 6)
)


def calcular_custo_com_sla(
    rota,
    matriz,
    limite_sla=50.0
):
    custo_total = 0.0
    penalidade = 0.0

    for i in range(len(rota) - 1):
        latencia_enlace = matriz[
            rota[i],
            rota[i + 1]
        ]

        custo_total += latencia_enlace

        if latencia_enlace > limite_sla:
            penalidade += 1000.0

    return custo_total + penalidade


rota_teste = np.array([
    0,
    1,
    2,
    3,
    4,
    5
])

custo_final = calcular_custo_com_sla(
    rota_teste,
    matriz_latencia
)


print("=" * 60)
print("EXERCÍCIO 2 - PENALIDADE DE SLA")
print("=" * 60)

for i in range(len(rota_teste) - 1):
    origem = rota_teste[i]
    destino = rota_teste[i + 1]
    latencia = matriz_latencia[origem, destino]

    status = (
        "VIOLA SLA"
        if latencia > 50
        else "OK"
    )

    print(
        f"{origem} -> {destino}: "
        f"{latencia:.2f} ms - {status}"
    )

print(
    f"\nCusto Total (Com Penalizações de SLA): "
    f"{custo_final:.2f} ms"
)
```
